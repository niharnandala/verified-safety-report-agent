```python
import pandas as pd
from data_prep import clean_data

raw_df = pd.read_excel("../data/Bisoprolol_icsr_sample_1068rows.xlsx")
cleaned = clean_data(raw_df)

reactions = cleaned["patient_reaction_reactionmeddrapt"]
outcomes = cleaned["patient_reaction_reactionoutcome"]

mismatch_count = 0
checked = 0
mismatch_rows = []

for i in range(len(cleaned)):
    r_val = reactions.iloc[i]
    o_val = outcomes.iloc[i]

    # skip rows where either side is totally missing, nothing to compare
    if pd.isna(r_val) or pd.isna(o_val):
        continue

    r_list = [x.strip() for x in str(r_val).split(",")]
    o_list = [x.strip() for x in str(o_val).split(",")]

    checked += 1

    # if the counts dont match, position by position pairing isnt safe for this row
    if len(r_list) != len(o_list):
        mismatch_count += 1
        mismatch_rows.append({
            "row": i,
            "reactions": r_list,
            "outcomes": o_list
        })

print("total rows checked:", checked)
print("rows where reaction count != outcome count:", mismatch_count)

for row in mismatch_rows:
    print(row)
```
